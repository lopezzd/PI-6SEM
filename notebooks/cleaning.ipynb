```python
import pandas as pd
import re
from pathlib import Path
```

```python
def repattern():
    """sumary_line
    
    Keyword arguments:
    argument -- description
    Return: return_description
    """ 
    path = Path('../data/raw')
    
    pattern = re.compile(r'^com\.samsung\.s?health\.(.+?)\.\d+$')
    
    for file in path.iterdir():
        if file.is_file():
            match = pattern.match(file.stem)
            
            if match:
                new_name = match.group(1)
                extention = file.suffix
                
                new_path = path / f"{new_name}{extention}"

                file.rename(new_path)
                print(f"Renamed: {file.name} -> {new_path.name}")
```

```python
repattern()
```
