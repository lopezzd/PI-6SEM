```python
import pandas as pd
import re
from pathlib import Path
```

```python
def repattern():
    """sumary_line
    
    Keyword arguments:
    argument -- description
    Return: return_description
    """ 
    path = Path('../data/raw')
    
    pattern = re.compile(r'^com\.samsung\.s?health\.(.+?)\.\d+$')
    
    for file in path.iterdir():
        if file.is_file():
            match = pattern.match(file.stem)
            
            if match:
                new_name = match.group(1)
                extention = file.suffix
                
                new_path = path / f"{new_name}{extention}"

                file.rename(new_path)
                print(f"Renamed: {file.name} -> {new_path.name}")
```

```python
def clean_and_save_data():
    raw_path = Path('../data/raw')
    processed_path = Path('../data/processed')
    processed_path.mkdir(parents=True, exist_ok=True)
    
    cols_to_drop = ['custom', 'create_sh_ver', 'modify_sh_ver', 'deviceuuid', 'pkg_name', 'datauuid']
    
    for file in raw_path.glob('*.csv'):
        try:
                first_line = f.readline()
                
            skip = 1 if "samsung" in first_line.lower() else 0:
            
            df = pd.read_csv(file, skiprows=skip, index_col=False, on_bad_lines='skip')
            
            df.columns = df.columns.str.strip()
            
            cols_to_drop_actual = [col for col in cols_to_drop if col in df.columns]
            df.drop(columns=cols_to_drop_actual, inplace=True)
            
            # Salva o arquivo ajustado
            clean_file_path = processed_path / file.name
            df.to_csv(clean_file_path, index=False)
            print(f"Limpo e salvo: {file.name} | Shape: {df.shape}")
            
        except Exception as e:
            print(f"Erro ao processar {file.name}: {e}")
```

```python
repattern()
```

```python
clean_and_save_data()
```
